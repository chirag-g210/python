### list vs numpy array

```python
py_list = [1,2,3]
print("python list multiplication", py_list*2)

np_array = np.array([1,2,3]) #element wise multiplication
print("python array multiplication", np_array*2)

import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n List operation time: ", time.time()-start)


start = time.time()
np_array = np.arange(1000000)*2
print("\n List operation time: ", time.time()-start)
```

# vector, matrix and tensor

```python
vector = np.array([1,2,3])
print("Vector:",vector)

matrix = np.array([[1,2,3],
                  [4,5,6]])
print("Matrix:",matrix)


tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]])
print("Tensor:",tensor)
```

## array reshaping


```python
arr = np.arange(12)
print("original array", arr)

reshaped = arr.reshape((3,4))
print("\n \n Reshaped array", reshaped)

# flatten (copy of original array , agr flatten perform krne ke bad kuch bhi krta hu array m jo muje flatten se mili h to original m kuch bhi chnges nhi honge!!)
flattened = reshaped.flatten()
print("\nflattened array", flattened)

# ravel (returns view , insted of copy , share same memory of original array so, chnges in array got after ravel , make chnges in original array!!)
raveled = reshaped.ravel()
print("raveled array", raveled)

```

# Transpose of array (matrices)

```python
arr = np.arange(12)
print("original array", arr)

reshaped = arr.reshape((3,4))
print("\n \n Reshaped array", reshaped)

# studied in maths............
transpose = reshaped.T
print("\n Transpose array", transpose)
```
